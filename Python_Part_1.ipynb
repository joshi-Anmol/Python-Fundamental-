Operators       
An operator is a symbol that tells Python to perform a specific computation on one
or more values (called operands)

````python
# Arithmetic Operators - used to perform math operations.

a = 5
b = 10
print(a + b) # Addition

print(a - b) # Subtraction

print(a * b) # Multiplication

print(a / b) # Division

print(a % b) # Modulo - remainder

print(a ** b) # Power

# Relational Operator - used to compare values 

a= 10, b = 18 
# Equal to
print("a == b:", a == b) # False

# Not equal to
print("a != b:", a != b) # True

# Greater than
print("a > b:", a > b) # False

# Less than
print("a < b:", a < b) # True

# Greater than or equal to
print("a >= b:", a >= b) # False

# Less than or equal to
print ("a <= b:", a <= b) # True

#Assingment Operator - Used to assign values to the variables 

# Simple assignment
x = 10
print("x =", x) # 10

x += 5 # equivalent to x = x + 5
print("x += 5 ->", x) # 15

x -= 3 # equivalent to x = x - 3
print("x -= 3 ->", x) # 12

x *= 2 # equivalent to x = x * 2
print("x *= 2 ->", x) # 24

x /= 4 # equivalent to x = x / 4
print("x /= 4 ->", x) # 6.0

x %= 4 # equivalent to x = x % 4
print("x %= 4 ->", x) # 2.0

x **= 3 # equivalent to x = x ** 3
print("x **= 3 ->", x) # 8.0


# Logical Operator - Used to combine Boolean Values 
x = 10
y = 20
z = 5

# AND
print(x > z and y > x) # True

# OR
print(x > y or y > z) # True

# NOT
print(not (x > y)) # True
````

# Operator Precedence in Python

**Operator precedence** defines the order in which Python evaluates operators when an expression contains multiple operators.

Operators with **higher precedence are evaluated before** operators with lower precedence.

### Example

```python
result = 10 + 5 * 2
```

The multiplication is performed first because `*` has higher precedence than `+`.

```text
10 + 5 * 2
10 + 10
20
```

Therefore:

```python
print(10 + 5 * 2)
# 20
```

## Python Operator Precedence — Highest to Lowest

| Priority | Operators                                                        | Description                                       |
| -------- | ---------------------------------------------------------------- | ------------------------------------------------- |
| 1        | `()`                                                             | Parentheses / grouping                            |
| 2        | `**`                                                             | Exponentiation                                    |
| 3        | `+x`, `-x`, `~x`                                                 | Unary plus, unary minus, bitwise NOT              |
| 4        | `*`, `/`, `//`, `%`                                              | Multiplication, division, floor division, modulus |
| 5        | `+`, `-`                                                         | Addition, subtraction                             |
| 6        | `<<`, `>>`                                                       | Bitwise shifts                                    |
| 7        | `&`                                                              | Bitwise AND                                       |
| 8        | `^`                                                              | Bitwise XOR                                       |
| 9        | `\|`                                                             | Bitwise OR                                        |
| 10       | `in`, `not in`, `is`, `is not`, `<`, `<=`, `>`, `>=`, `==`, `!=` | Comparisons, membership, identity                 |
| 11       | `not`                                                            | Logical NOT                                       |
| 12       | `and`                                                            | Logical AND                                       |
| 13       | `or`                                                             | Logical OR                                        |
| 14       | `if ... else`                                                    | Conditional expression                            |
| 15       | `lambda`                                                         | Lambda expression                                 |


### Example

```python
result = 2 + 3 * 4
# 14
```

`*` has higher precedence than `+`, so multiplication happens first.

### Parentheses

Parentheses can be used to explicitly control the order:

```python
(2 + 3) * 4
# 20
```

**Important:** Operators with the same precedence generally follow **left-to-right associativity**, except `**`, which associates **right-to-left**.

```python
10 - 3 - 2      # (10 - 3) - 2 = 5
2 ** 3 ** 2     # 2 ** (3 ** 2) = 512
```
