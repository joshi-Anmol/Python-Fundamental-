Concepts : Object Oriented Programming in Python 


Object-oriented programming (OOP) in Python is a programming paradigm centered around organizing code into objects: bundles of data (attributes) and behaviour (methods). It helps us structure programs in a way that is modular, reusable, and easier to maintain.

OOP models software as a collection of interacting objects, similar to how we think about real-world entities. Each object represents something meaningful in our program - such as a user, a car, a bank account, or an enemy in a video game.

## Encapsulation 

Encapsulation is the bundling of data (variables) and methods (functions) that      
operate on that data into a single unit (a class), along with controlling access to that         
data. This is done to protect the data from accidental or unauthorized modification.

```python
# Wrapping data in a single unit i.e; a class is called encapsulation 

class Bankaccount:
    def __init__(self, name, balance, ID):
        self.name = name # public members
        self._balance = balance # _ with a variable name makes it a protected member 
        self.__ID = ID # Private member of the class , by conventioin shouldn't be used outside the class 

    def get_ID(self): # Getter method 
        return self.__ID

    def set_ID(self , new_ID): # setter  method 
        self.__ID = new_ID


holder1= Bankaccount("Virat",1_000_000,1)
print(f"Accessing all the details for account holder: \n name: {holder1.name}, balance: {holder1._balance}")

# Now to manipulate the private variables we use getters and setters methods 

print(holder1.get_ID())
holder1.set_ID(4)
print(holder1.get_ID())
```

## NOTE Name mangling

public      → normal access     
protected   → internal-use convention       
private     → name mangling + stronger internal-use convention      

# Inheritance
Inheritance is where one class (child) acquires the properties and behaviors (variables + methods) of another class (parent).   
The class whose properties are inherited - Parent / Base / Superclass   
The class that inherits - Child / Derived / Subclass    

```python
class Employee: # parent class
    start_time = "9AM"
    end_time = "5PM"

class Teacher(Employee): # child class
    def __init__ (self, subject):
        self.subject = subject


t1 = Teacher("Data Science")
print(t1.subject, t1.start_time, t1.end_time)
```

```python
# Types of Inheritance 

# Single Inheritance: Parent → Child

class Parent:
    def display(self):
        print("Parent class")

class Child(Parent):
    pass

c = Child()
c.display() # Output: Parent class


# Multi level Inheritance 

class parent_class:
    def get_const():
        print("Getter function works")

class child_class(parent_class):
    def geto():
        print("lets goooooooo")

class derived(child_class):
    pass 

child_object = derived()

derived.geto()
derived.get_const()

# Multiple Inheritance 

class Teacher:
    def init (self, salary):
        self.salary = salary

class Student():
    def init (self, gpa):
        self.gpa = gpa
        
class TA(Teacher, Student):
    def __init__ (self, name, salary, gpa):
        super(). init (salary) # call parent constructor
        Student. init (self, gpa) # call parent constructor
        self.name = name

ta = TA("Rahul", 50_000, 7.5)
print(ta.name, ta.salary, ta.gpa)
```

# Abstraction   
Abstraction is hiding unnecessary implementation details and showing only the essential features to the user.       
We implement abstraction with abstract classes & abstract methods.


Abstract Class      
An abstract class in Python is one which:       
• Cannot be instantiated    
• Can contain normal + abstract methods     
• Usually acts as a blueprint for child classes     

```python
from abc import ABC , abstractmethod

class animal(ABC):
    @abstractmethod
    def sound(self):
        pass 

class lion(animal):
    def sound(self):
        print("Roar!")

class Cat(animal):
    def sound(self):
        print("Meow!")

lion1= lion()
lion1.sound()
```

## Note
How abstraction is different from data hiding   
DATA hiding means hiding data for external excess where as abstraction means hiding unnecessary implementation details and shwocasing functionality only 


# Polymorphism  

Polymorphism is the ability of a single function, operator, or object to behave
differently based on the context. (“poly” = many, “morph” = forms)      

The idea is that:   
• Same method name - works differently for different objects        
• Same operator - behaves differently depending on operand types    


```python
# Example 
print(1 + 2) # adds 2 numbers
print("1" + "2") # concatenates 2 strings
```

1. Function Overriding (or Method Overriding):      
• When a child class provides its own version of a method that already exists in
the parent class (Both methods should have same name)       
• Type of Runtime Polymorphism (dynamic binding)        
• Child method takes precedence over parent method.     


2. Duck Typing      
• Works on the idea: “If it looks like a duck and quacks like a duck, it must be a
duck.”      


```python
# Function Overriding 

class animal:
    def sound(self):
        print("Makes sound!!")

class Dog(animal):
    def sound(self):
        print("Bark bitch!!")

Pomerian = Dog()
Pomerian.sound()

#  Duck Typing 

class Dog:
    def speak(self):
        print("Bark")

class Cat:
    def speak(self):
        print("Meow")

class Robot:
    def speak(self):
        print("Beep Boop")

def make_it_speak(entity):
    entity.speak()  # doesn't care about type


d = Dog()
c = Cat()
r = Robot()

for e in [d, c, r]:
    make_it_speak(e)
```
