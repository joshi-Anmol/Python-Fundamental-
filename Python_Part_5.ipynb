## Files input output 
File I/O (Input/Output) means reading data from files and writing data to files 
using Python. 
Python provides built-in functions to work with files.

```python
# opening a file 
file = open("sample.txt","r")

data = file.readline()
print(data)

file.close()
```

```python
# readlines usage 
file = open("sample.txt","r")

data = file.readlines()
print(data)

file.close()
```

```python
# common modes in file i/o

# writing 
file = open("Sample.txt","w")

file.writelines("Hey their this is the usage of file i/o \n and i am really sorry for your loss")

file.close()

# append 

file = open("Sample.txt","a")
file.writelines("\nNow i am appending these line as the structure")
file.close()
```

```python
# With keyword 

with open("sample.txt","r") as f :
    data = f.readlines()
    print(data)
```

```python
# Practice problem to find a specific data into a txt file 

data = True 
line = 1
word ="line"

with open("Sample.txt","r") as f:
    while data: # the loop with until its returning valid data 
        data = f.readline()

        if word in data:
            print(f"The word {word} is found at line {line}")
            break 
        line+= 1
```

```python
# List comprehensions : ist Comprehension is a short and elegant way to create lists in Python.It replaces long `for` loops with one-line expressions.

print("Task: Create a list of numbers form 1 to 5 ")
my_list = [n for n in range(1,6)]
print(my_list)

# list comprehensions with conditions 
print("Task: Even numbers form 1 to 10")
Even_call =[x  for x in range(1,11) if x%2==0]

# List comprehensions with if else conditions 
print("Task: label numbers as even or odd ")
even_odd = ["even" if x%2==0 else "odd" for x in range(1,11)]
print(even_odd)

print("Task: Convert every negative integer as 0")
list_given = [-1,-22,23,3,42,0,-2456,23,0 ]
zero_convertor =[0 if x < 0 else x for x in list_given ]
print(zero_convertor)
```

# JSON 
`Javascript object notataion` is a light weight data format used to exchaneg data between programs, APIs, websites, etc.        
It is very similiar to Python dictionary 

```python
import json 
"""
JSon has four functions load(), dump(), loads(), dumps()

"""
# Converting Python data into JSON data  
raw_json_str_data = '{"name" : "Anmol","Age" : 22, "Designataion" : "Data Scientst"}'

# converting JSON string into a dictionary `python object`` we use loads() function 
print(type(raw_json_str_data))
python_object = json.loads(raw_json_str_data)
print(type(python_object))
print(python_object)

# conerting an oython dictionary into a json string 
raw_python_dict = {
    "name":"Anmol" ,
    "age":22,
    "city":"Haldwani",
    "designation":"AI Engineer"
                  }
print(type(raw_python_dict))
json_string = json.dumps(raw_python_dict)
print(type(json_string))
print(json_string) 

#  when dealing with files not data 

# reading data from a JSON file 
with open("Data_format.json" ,"r")as f :
    python_dict = json.load(f)
    print(python_dict)
    print(type(python_dict))

# changing the data into json file 
with open("Data_format.json","w") as f :
    json.dump(raw_python_dict, f , indent=4, sort_keys=True)
```
